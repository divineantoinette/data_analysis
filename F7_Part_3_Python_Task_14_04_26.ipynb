```python
#Import Pandas into Google Colab
import pandas as pd
```

```python
#Save the ‘F7Data’ file in Google Drive and load into Google Colab
new_data = pd.read_csv("/content/drive/MyDrive/Think Employment Course Data/Python/F7Data.csv")
new_data
```

```python
copied_data = new_data.copy()
copied_data
```

```python
#Find the data types
copied_data.info()
```

```python
#Rename the ‘County’ column to ‘Country’
rename_country = {"County" : "Country"}

copied_data.rename(columns = rename_country, inplace = True) #Tell Python you want to rename the columns, in what variable it is stored, and to put it into place and label it true to overried the names/execute the command
copied_data
```

```python
#Correct the spelling of Charlotte
copied_data["Name"] = copied_data["Name"].replace(["Charlote"], "Charlotte")
copied_data["Name"]
```

```python
copied_data.info()
```

```python
#Find the average (mean) age
mean_age = format(copied_data["Age"].mean(), ".1f")
mean_age
```

```python
#Replace the missing ages with the average
copied_data.fillna({"Age" : mean_age}, inplace=True) #This replaces the null values with the median age
#copied_data.fillna({"Age" : "37.2"}, inplace=True) #This is an alternative code for the STRING VALUE (to achieve the same result)
copied_data
```

```python
copied_data.info() #This shows that after changing values, the data type of age changes
```

```python
#To fix this, we can force the data type back to the orignial
copied_data["Age"] = copied_data["Age"].astype(float) #The .astype() command forces the data type (as float in this instance)
copied_data.info()
```

```python
#Find the average (mean) height (no decimal places)
mean_height = format(copied_data["Height"].mean(), ".1f")
mean_height
```

```python
#Replace the missing heights with the average
copied_data.fillna({"Height" : mean_height}, inplace=True) ##This replaces the null values with the median height
#copied_data.fillna({"Height" : "174"}, inplace=True)
copied_data
```

```python
copied_data.info() #This shows that after changing values, the data type of height changes
```

```python
#To fix this, we can force the data type back to the orignial
copied_data["Height"] = copied_data["Height"].astype(float) #The .astype() command forces the data type (as float in this instance)
copied_data.info()
```

```python
#Delete (drop) the weight column
copied_data.drop(columns = "Weight", inplace = True)
copied_data
```

```python
#Set Liam’s and James’ gender to M
copied_data.loc[9] = {"Name" : "Liam",	"Age" : 37.2,	"Height" : 175.0,	"Gender" : "M", "Country" : "Scotland"}
copied_data.loc[15] = {"Name" : "James",	"Age" : 60.0,	"Height" : 188.0,	"Gender" : "M", "Country" : "Scotland"}
copied_data
```

```python
#Delete the row containing William’s details
copied_data.drop(index = [18], inplace = True)
copied_data
```

```python
#Find the total age
total_age = copied_data ["Age"].sum()
total_age
```

```python
#Find the minimum height
min_height = copied_data ["Height"].min()
min_height
```

```python
#Find maximum height
max_height = copied_data ["Height"].max()
max_height
```

```python
#Replace the incorrect spellings for the countries England, Scotland and Wales
copied_data["Country"].unique()
```

```python
#Replace the incorrect spellings for the countries England, Scotland and Wales
copied_data["Country"] = copied_data["Country"].replace(["English", "Eng"], "England")
copied_data["Country"] = copied_data["Country"].replace(["Scottland"], "Scotland")
copied_data["Country"] = copied_data["Country"].replace(["WAles"], "Wales")
copied_data["Country"]
```

```python
#Check there are no null values left in any column using a function
copied_data.isnull().sum()
```
