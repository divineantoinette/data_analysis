```python
#Create a copy of the dataset, so that the original data set is not altered and in case any changes made are not correct

copied_data = new_data.copy()
copied_data
```

```python
#Use a dictionary to change the column names
#A dictionary is an organised locker system, everything is stored with a unique name, it is used to store data values and types
#To call on a dictionary, you need to use curly braces

names = {"Distance" : "Dis", #Use quotation marks as the objects are string
         "Pulse" : "Pul",  #Use commas as a seperator
         "Maxpulse" : "Max", #The original column name is on the left, the new column name is after the colon
         "Calories" : "Cal"}

copied_data.rename(columns = names, inplace = True) #Tell Python you want to rename the columns, in what variable it is stored, and to put it into place and label it true to overried the names/execute the command
copied_data
```

```python
copied_data.isnull().sum() #This command is a quicker way to show you the null values in each column
```
