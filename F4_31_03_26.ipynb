Iteration Types (Loops)

```python
#Create a FOR loop: calculate the sum of the first 5 numbers

total = 0

for i in range(1,6):
  total = total + i

print(total)

#Loop i value total
# 1       1     1 #i + total value = total column
# 2       2     3
# 3       3     6
# 4       4     10
# 5       5     15
```

```python
#Create a WHILE loop: calculate the sum of the first 5 numbers
#i, x, and j can be used interchangeably and there is no difference in the code outcome

total = 0
i = 1
while i <=5:
  total = total + i
  i = i + 1

print(total)

# Iteration	        i (before adding)	    total = total + i	    i after i + = 1

# 1	                1	                    0 + 1 = 1	                  2

# 2	                2	                    1 + 2 = 3	                  3

# 3	                3	                    3 + 3 = 6	                  4

# 4	                4	                    6 + 4 = 10	                5

# 5	                5	                    10 + 5 = 15	                6



#After this iteration, where i = 6, i is not less than (<) and not equal to (=) 5, therefore the condition is False and this means that the loop will stop counting
```

Boolean Operators (To compare values to another)

```python
a = 64
b = 23
c = 84

print(a > b) # > means greater or more than
print(a < b) # < means less than
print(a >= b) # >= means more than or equal to
print(a <= c) # <= means less than or equal to

print(b==c) # == means equal in Python
print(b != c) # ! means not

#A boolean will print as either True or False, no other value
```

Logical Operators

```python
#AND is only True if everything is True where all conditions must be met together (every condition has to be true)

age = int(input("Enter your age: "))

if age >= 0 and age < 18 :
  print("Child")

elif age <0:
  print("Error, you've entered in invalid age, the number must be between 0-99")

else:
  print("Adult")
```

```python
#OR is only True if at least on condition is (if one of the conditions are true

num = int(input("Enter a number: "))

if num <0 or num >50:
  print("The number is not between 0 and 50")

else:
  print("The number is between 0 and 50")
```

```python
#ELIF allows you to check if multiple statements are True (if the previous conditions were not true, then try this condition)
#What if we need more than two options? We will use an elif statement (else if)

age = int(input("Enter your age: "))

if age >= 18:
  print("Adut")

elif age >= 13 and age <18:
  print("Teenager")

else:
  print("Child")
```

```python
# Task 3
# Create a calculator
# Ask the user to enter the first number

first_number = float(input("Enter the first number: "))

# Ask the user which operator they would like to use e.g. +, -, *, /

equation = input("What equation would you like to carry out? (+, -, *, /): ")

# Ask the user to enter the second number

second_number = float(input("Enter the second number: "))

if equation == "+": # Addition
  print(first_number + second_number)

elif equation == "-": # Subtraction
  print(first_number - second_number)

elif equation == "*": # Multiply
  print(first_number * second_number)

elif equation == "/": # Division
  print(first_number / second_number)

else:
  print("You have not entered a correct operator")
```
