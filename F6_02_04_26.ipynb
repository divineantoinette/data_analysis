How to Clean Data

```python
#How to clean data (where errors were made on purpose)
#Columns are lists, meaning it requires square brackets

sales_data["Country"].unique() #By using parentheses, we are asking python to do something that is not specified to one part of the dataset
                              #The .unique command shows unique values
```

```python
#Replace the values for Sweden
sales_data["Country"] = sales_data["Country"].replace(["SwedeN","SWEDEN"], "Sweden") #The .replace command replaces values
                                                                                    #The square brackets are used to refer to a list
```

```python
sales_data["Country"].unique()
```

```python
#Replace the values for France

sales_data["Country"] = sales_data["Country"].replace(["FrancE","FrAnce", "france"], "France")
```

```python
#Replace the values for United Kingdom

sales_data["Country"] = sales_data["Country"].replace(["United kingdom"], "United Kingdom")
```

```python
#Replace the values for Italy

sales_data["Country"] = sales_data["Country"].replace(["ITAly"], "Italy")
```

```python
sales_data["Country"].unique()
```

```python
#Check the amount of countries
num_of_countries = len(sales_data["Country"].unique()) #LEN is short for length, length is a counter
num_of_countries
```

```python
#Check the amount of cities
num_of_cities = len(sales_data["City"].unique())
num_of_cities
```
