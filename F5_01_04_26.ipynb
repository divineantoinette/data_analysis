```python
new_data.describe() #This commmand gives us a description of the numerical statistics ONLY

#If there are any text values, it would leave them out
# Count - counts up the number of values and helps us find missing values
# Mean - give us a mean average value by adding all the values and dividing by total number of values
# Standard deviation (otherwise known as std) - measures the spread of values
# Minimum - gives us the minimum (lowest) value
# 25% - gives us the first quartile
# 50% - gives us the second quartile
# 75% - gives us the third quartile
# Max - gives us the maximum (highest) value
```
