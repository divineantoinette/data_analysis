```python
# Joining strings together
first_name = "Divine"
middle_name = "Antoinette"
last_name = "Brabbs"
print(first_name + last_name) # The addition sign will join these variables together
print(first_name,last_name) # A comma is a seperator that makes a space between the string/words
print(first_name + " " + last_name) # You can also join string with a delimieter (requires quotation marks around it) and include a space
print(first_name + middle_name + last_name)
print(first_name, middle_name, last_name)
print(first_name + " " + middle_name + " " + last_name)
print(f"{first_name} {middle_name} {last_name}") # Creating an f-string uses {} curly braces and whatever is between the curly braces is added into the print command
print(f"{first_name}-{middle_name}-{last_name}")
```

```python
# In class activity
name = "Divine"
age = 27
region = "Manchester"
course = "Data Analytics"
print(name)
print(age)
print(region)
print(course)
#OR
print(name,age,region,course)
# Hello my name is [name], my age is [age]. I am from [region] and the course I'm studying is called [course].
# Method 1: Using f-strings
# Method2: Standard way (+)

# Method 1
print(f"Hello my name is {name}, my age is {age}, I am from {region}, and the course I'm studying is called {course}.")
# Method 2
print("Hello my name is" + name + ", my age is " + str(age) + ". I am from" + region + "and the course I'm studying is called" + course + ".")
# You can force variables to be a different data type, for example age turned into str(age)
```

```python
# Joining example
name = "Divine"
age = 27
print("My name is " + name + " and my age is " + str(age))
# TypeError: can only concatenate str (not "int") to str
# It cannot only join string up to string and not to integers, therefore you are required to convert age into string by str(age) or "27"
print(f"My name is {name}, my age is {age}.")
```

```python
# Data types
name = "Divine" # STR is text
age = 27 # INT is a whole number
height = 1.57 # FLOAT is a floating point or a decimal
course = False # BOOL is a boolean, which can only and always be true or false
mylist = [11, "Anthony", 1.56] # LIST as a variable that can contain multiple different data types (otherwise known as an array)
print(type(name)) # This will print the data type of each variable
print(type(age))
print(type(height))
print(type(course))
print(type(mylist))
```
