NumPy (Numerical Python) &

MatPlotLib (Mathematical Plot Library) which allows you to plot charts and graphs

```python
#When completing mathematical work, use the numpy library
import numpy as np
import matplotlib.pyplot as plt #The matplotlib library is vast, so we only need to import a section which is pyplot
import os #This allows us to interact with files on our computer
```

```python
#Basic structure of a plot
x_axis = np.array([0,10]) #X and y axis are variables, where we will store as an array to plot a graph
y_axis = np.array([0,10]) #X and y axis need the same amount of numbers in each axis to work
plt.plot(x_axis, y_axis) #Create a line graph using .plot
plt.show() #Is the equivalent to the print command
```

```python
#Basic structure with addedd coordinates
x_axis = np.array([2,4,6,8,10]) #X and y axis are variables, where we will store as an array to plot a graph
y_axis = np.array([0,1,21,35,60]) #X and y axis need the same amount of numbers in each axis to work
plt.plot(x_axis, y_axis, "bo") #bo is short for blue objects (there are 8 colours, e.g. r for red, y for yellow, b for blue, c for cyan, m for magenta, y for yellow, b for black and w for white)
plt.show() #Is the equivalent to the print command
```

```python
#Basic structure with added coordinates
x_axis = np.array([2,4,6,8,10])
y_axis = np.array([0,1,21,35,60])
plt.plot(x_axis, y_axis, "g+") #g+ stands for green + signs (there are X shapes, e.g. o = objects, p = pentagons, s = squares, + = plus signs, * for star, x for X, D for diamond, d for a thin diamond, etc.)
plt.show()
```

```python
#Basic structure with added coordinates
x_axis = np.array([2,4,6,8,10])
y_axis = np.array([0,1,21,35,60])
plt.plot(x_axis, y_axis, linestyle = "dashed") #linestyle = "dashed" can change the style of the line to dashes
plt.show()
```

```python
#Naming the line graph and the axis
x_axis = np.array([2,4,6,8,10])
y_axis = np.array([0,1,21,35,60])
plt.plot(x_axis, y_axis)
plt.xlabel("Pulse")
plt.ylabel("Calories")
plt.title("Exercise Readings")
plt.show()
```

```python
#Create a bar chart (used for categorical data)
x_axis = np.array(["Pen", "Pencil", "Ruler", "Eraser"])
y_axis = np.array([8, 4, 4, 1])
plt.bar(x_axis,y_axis, color = "Green") #Change to a bar graph .bar
plt.xlabel("Item")
plt.ylabel("Quantity")
plt.title("Stationary")
plt.show()
```

```python
#Create a bar chart (used for categorical data)
x_axis = np.array(["Pen", "Pencil", "Ruler", "Eraser"])
y_axis = np.array([8, 4, 4, 1])
plt.bar(x_axis,y_axis, color = ["Black", "Blue", "Yellow", "Magenta"])
plt.xlabel("Item")
plt.ylabel("Quantity")
plt.title("Stationary")
plt.show()
```

```python
#Create a scatter chart
x_axis = np.array([2,4,6,8,10])
y_axis = np.array([0,1,21,35,60])
plt.scatter(x_axis, y_axis) #Change to a scatter graph .scatter
plt.xlabel("Pulse")
plt.ylabel("Calories")
plt.title("Exercise Readings")
plt.savefig("my_scatter_chart.png")
plt.show()
```
