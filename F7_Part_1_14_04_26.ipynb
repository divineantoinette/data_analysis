```python
import pandas as pd
```

```python
#Load the data
num_data = pd.read_csv("/content/drive/MyDrive/Think Employment Course Data/Python/pandas data class example.csv")
num_data
```

```python
#Filtering the data distances greater than 200
num_data.loc[num_data["Distance"] > 200]
```

```python
#Filtering the data distances less than or equal to 100
num_data.loc[num_data["Distance"] <= 100]
```

```python
#Filtering the data calories more than 160
num_data.loc[num_data["Calories"] > 160]
```

```python
#Calculate the total distance (when you think of total, think of the sum aggregation)
total_distance = num_data ["Distance"].sum()
total_distance
```

```python
#Calculate the minimum distance
min_distance = num_data ["Distance"].min()
min_distance
```

```python
#Calculate the maximum distance
max_distance = num_data ["Distance"].max()
max_distance
```

```python
#Calculate the mean distance (this is used to work out the missing values in a dataset)
mean_cal = num_data ["Calories"].mean()
mean_cal
```

```python
#Formatting the mean (moving the decimal places)
mean_cal = format(num_data ["Calories"].mean(), ".2f")
mean_cal
```
