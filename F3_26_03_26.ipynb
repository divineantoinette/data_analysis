```python
# Calculate the amount owed on a loan (workout the percentage)
loanAmount = int(input("Enter the loan amount: "))
interestRate = int(input("Enter the interest rate: "))
duration = int(input("Enter the duration: "))
totalInterest = loanAmount * interestRate/100 * duration/12
totalAmount = loanAmount + totalInterest

print(f"The total amount owed is {totalAmount}") #Create a new variable to print

#OR

print(f"The total amount owed is {totalInterest + loanAmount}") #Include the calculation in the curly braces
```

```python
#Ask the user for a savings amount, interest rate and number of years, then calculate the final savings using simple interest
current_savings_amount = float(input("Enter the savings you would like to place into a savings account: £"))
interest_rate = float(input("The interest percentage is: "))
duration = int(input("The number of years the interest is provided for: "))

total_interest = current_savings_amount * interest_rate/100 * duration/12
final_savings = current_savings_amount + total_interest

print(final_savings)
print(f"The total savings amount at the end of {duration} months will be: £{current_savings_amount + total_interest}")

#Convert percentage to decimal
interest_rate = interest_rate/100
```

```python
#Ask for a restaurant bill, tip percentage and number of people, then calculate how much each person should pay
restaurant_bill = float(input("Enter the price of the restaurant bill: "))
tip_percentage = float(input("Enter the tip percentage suggested by the restaurant: "))
number_of_people = int(input("Enter the number of people who attended the meal: "))

total_restaurant_bill = restaurant_bill * tip_percentage/100 + restaurant_bill
cost_per_person = total_restaurant_bill / number_of_people

print(cost_per_person)
print(f"Each person should pay £ {cost_per_person} each")
```

```python
#Ask for a better stake and odds, then caculate and print the total return and profit
stake = float(input("What is the stake? "))
odds = float(input("What are the odds? "))

total_return = stake * odds
total_profit = total_return - stake

print(total_return)
print(total_profit)
print(f"The total return is £ {total_return}")
print(f"The total profit is £ {total_profit}")
```
